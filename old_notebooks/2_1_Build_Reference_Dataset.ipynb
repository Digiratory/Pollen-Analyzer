# Reference dataset

This notebool is intended to prepare reference dataset for trainig of OSC model. 

We cannot use the preliminary segmented images from `objects_clean` subset because the segmentation model introduce additional artifacts which affect the quality of OSC classification model prediction.

```python
import os
from glob import glob
import math
from skimage import io
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
```

```python
from Instance_Segmentation import SegmentationModelWrapper
from common.configuration import ModelPath, DatasetPath, species_invasive, species_native
```

```python
segmentation_model = SegmentationModelWrapper(ModelPath.instance_segmentation)
```

```python
segmentation_input_dir = os.path.join(DatasetPath.raw_sample_images, "Acer")
# segmentation_input_dir = r"Dataset\pollen_dataset_2025_02_05\samples\Alnus"
images = glob(os.path.join(segmentation_input_dir, "*.jpg"))

# Plot some samples
samples = images[0:9]
n = math.ceil(len(samples) ** (1/2))
fig, axs = plt.subplots(nrows=n, ncols=n, figsize=(n*2, n*2))
for ax, img_path in zip(axs.flatten(), samples):
    image_rgb = io.imread(img_path)
    ax.imshow(image_rgb)
    ax.set_axis_off()
    ax.set_title(os.path.basename(img_path))
for ax in axs.flatten()[len(samples):]:
    ax.set_visible(False)
fig.suptitle('Segmentation Input', fontsize=14)
plt.tight_layout()
plt.show()
```

## Build dataset

Iterate over folders and build the same structure in a target folder

```python
species_dirs = glob(os.path.join(DatasetPath.raw_sample_images, '*'))
species_dirs = list(filter(lambda f: os.path.isdir(f), species_dirs))
species_dirs
```

```python
for species_dir in tqdm(species_dirs):
    images = glob(os.path.join(species_dir, "*.jpg"))
    for image_path in tqdm(images):
        segmentation_model(image_path=image_path, output_dir=DatasetPath.reference_images)
```
