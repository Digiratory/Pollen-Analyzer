## Train model

To train model you can also execute the script `Instance_Segmentation/train_model.py`as following

```python
python Instance_Segmentation/train_model.py
```

The module's aid is to configure and train the model, as well as saving the metrics, sample images during the way.

- noise_dir - source path to Noise(background) images
- pollen_dir - path to a directory where pollen samples are held
- model_path - path, where trained model weights will be saved
- image_save_dir - path, where processed generated images will be saving during the training phase
- log_dir - path to a directory where training metrics will be saved

````python
from common.configuration import DatasetPath, ModelPath
````

````python
noise_dir = DatasetPath.noise_bg_images # 'Data/Noise_40X_95'
pollen_dir = DatasetPath.objects_clean
model_path = ModelPath.instance_segmentation #'Save/models/InSegModel_fb'
log_dir = 'Save/metrics/log_fb'
image_save_dir = 'Save/images/images_fb'
````

NOTE: during research, the invasive species has been removed from the dataset folder manually.

````python
import Instance_Segmentation.train_model as train_model
if DO_TRAIN_FLAG:
    train_model.main(noise_dir, pollen_dir, model_path, log_dir, image_save_dir)
````
